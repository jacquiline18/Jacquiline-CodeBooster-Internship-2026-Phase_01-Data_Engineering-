```python
def run_query(sql, description=""):
  conn = sqlite3.connect(':memory:')
  df.to_sql('students', conn, if_exists='replace', index=False)
  if description:
    print(f"\n{'='*55}")
    print(f"{description.upper()}")
    print(f"{'='*55}")
  result=pd.read_sql_query(sql, conn)
  print(result)
  conn.close()
  return result
```

```python
query1= """SELECT name, department, math_score
FROM
students ORDER
BY math_score DESC LIMIT 5"""
```

```python
run_query(query1, description="Top 5 Students by Math Score")
```

```python
query2= """SELECT name, department, attendance_percentage FROM students WHERE attendance_percentage > 90 ORDER BY attendance_percentage DESC"""
```

```python
run_query(query2, description="Students with High Attendance (>90%)")
```

```python
import pandas as pd
import sqlite3

# Ensure df is defined for this cell's operations, using the full path to the file
df = pd.read_csv('/content/student_performance.csv')

# Establish a connection for this cell's operations
conn_local = sqlite3.connect(':memory:')

# Load the main DataFrame 'df' into the 'students' table in this connection
df.to_sql('students', conn_local, if_exists='replace', index=False)

# Define query5 (assuming it's the intended query for department data)
query5 ="""
 SELECT
  department,
  COUNT(*) AS student_count,
  ROUND(AVG(math_score),2) AS avg_math_score,
  ROUND(AVG(science_score),2)AS avg_science,
  ROUND(AVG(programming_score),2) AS avg_programming
FROM STUDENTS
GROUP BY department
ORDER BY avg_math_score DESC
"""

# Get the department data using pd.read_sql_query against the local connection
# This directly creates the DataFrame 'dept_df'
dept_df = pd.read_sql_query(query5, conn_local)

# Now, save this dept_df to a new table 'departments' in the same local connection
dept_df.to_sql('departments', conn_local, if_exists='replace', index=False)

print("Created 'departments' table:")
print(dept_df.to_string(index=False))

# Close the local connection
conn_local.close()
```
